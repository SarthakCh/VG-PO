```python
import os
import pandas as pd
import numpy as np

# Define the file path (adjust as needed based on your uploaded dataset structure)
uploaded_file_path = "prices.csv"  # Adjust if needed
output_directory = "Desktop/VG portfolio optimization/PO implement/dcmppln-main/dcmppln"

```

```python
 
def process_russell3000_dataset(file_path):
    """
    Load and preprocess the Russell 3000 dataset to compute correlation, covariance, and returns data.

    Args:
        file_path (str): Path to the dataset CSV file.

    Returns:
        np.ndarray: Correlation matrix.
        np.ndarray: Covariance matrix.
        np.ndarray: Returns data.
    """
    # Load the dataset
    data = pd.read_csv(file_path)
    print(f"Loaded dataset with {data.shape[0]} rows and {data.shape[1]} columns.")

    # Pivot dataset to organize stock prices by date and ticker
    stock_data = data.pivot(index='Date', columns='Ticker', values='Close').dropna()

    # Compute daily percentage returns
    returns_data = stock_data.pct_change().dropna().values

    # Compute correlation and covariance matrices
    correlation_matrix = np.corrcoef(returns_data, rowvar=False)
    covariance_matrix = np.cov(returns_data, rowvar=False)

    return correlation_matrix, covariance_matrix, returns_data

# Step 3: Save the matrices in `.npy` format
def save_matrices(correlation_matrix, covariance_matrix, returns_data, output_dir):
    """
    Save correlation_matrix, covariance_matrix, and returns_data as .npy files in the specified directory.

    Args:
        correlation_matrix (np.ndarray): Correlation matrix.
        covariance_matrix (np.ndarray): Covariance matrix.
        returns_data (np.ndarray): Returns data.
        output_dir (str): Directory to save the .npy files.
    """
    os.makedirs(output_dir, exist_ok=True)
    np.save(os.path.join(output_dir, "correlation_matrix.npy"), correlation_matrix)
    np.save(os.path.join(output_dir, "covariance_matrix.npy"), covariance_matrix)
    np.save(os.path.join(output_dir, "returns_data.npy"), returns_data)
    return [
        os.path.join(output_dir, "correlation_matrix.npy"),
        os.path.join(output_dir, "covariance_matrix.npy"),
        os.path.join(output_dir, "returns_data.npy"),
    ]

# Execute the processing and saving
try:
    correlation_matrix, covariance_matrix, returns_data = process_russell3000_dataset(uploaded_file_path)
    output_files = save_matrices(correlation_matrix, covariance_matrix, returns_data, output_directory)
    output_files
except Exception as e:
    str(e)
```

```python
import pandas as pd
import numpy as np
import os

def process_and_save_russell3000_data(file_path, output_dir):
    """
    Process the Russell 3000 dataset and save correlation, covariance, and returns as .npy files.

    Args:
        file_path (str): Path to the CSV file.
        output_dir (str): Directory where the .npy files will be saved.
    """
    # Load the dataset
    data = pd.read_csv(file_path, parse_dates=['date'])  # Adjust for the 'date' column
    print(f"Loaded dataset with {data.shape[0]} rows and {data.shape[1]} columns.")

    # Pivot dataset to organize stock prices by date and symbol
    stock_data = data.pivot(index='date', columns='symbol', values='close')

    # Check for missing data
    print("Missing values before dropna:", stock_data.isnull().sum().sum())

    # Handle missing data
    stock_data = stock_data.dropna(axis=1, how='any')  # Drop columns with any missing values
    print("Missing values after dropna:", stock_data.isnull().sum().sum())

    # Compute daily percentage returns
    returns_data = stock_data.pct_change().dropna().values

    # Check returns data for NaN
    print("Returns data shape:", returns_data.shape)
    print("NaN in returns data:", np.isnan(returns_data).sum())

    # Compute correlation and covariance matrices
    correlation_matrix = np.corrcoef(returns_data, rowvar=False)
    covariance_matrix = np.cov(returns_data, rowvar=False)

    # Handle potential NaN values in matrices
    if np.isnan(correlation_matrix).any():
        print("Warning: NaN detected in correlation matrix.")
    if np.isnan(covariance_matrix).any():
        print("Warning: NaN detected in covariance matrix.")

    # Create output directory if it doesn't exist
    os.makedirs(output_dir, exist_ok=True)

    # Save the results as .npy files
    np.save(os.path.join(output_dir, 'russell3000_correlation.npy'), correlation_matrix)
    np.save(os.path.join(output_dir, 'russell3000_covariance.npy'), covariance_matrix)
    np.save(os.path.join(output_dir, 'russell3000_returns.npy'), returns_data)

    print("Files saved successfully:")
    print(f" - Correlation matrix: {os.path.join(output_dir, 'russell3000_correlation.npy')}")
    print(f" - Covariance matrix: {os.path.join(output_dir, 'russell3000_covariance.npy')}")
    print(f" - Returns: {os.path.join(output_dir, 'russell3000_returns.npy')}")

# Usage example
file_path = 'prices.csv'  # Path to your dataset
output_dir = 'Desktop/VG portfolio optimization/PO implement'  # Directory to save .npy files
process_and_save_russell3000_data(file_path, output_dir)
```
